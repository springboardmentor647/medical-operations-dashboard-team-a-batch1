```python
#convert data  columns
patients["date_of_birth"] = pd.to_datetime(patients["date_of_birth"])

patients["registration_date"] = pd.to_datetime(
    patients["registration_date"])

appointments["appointment_date"] = pd.to_datetime(
    appointments["appointment_date"])

treatments["treatment_date"] = pd.to_datetime(
    treatments["treatment_date"])

billing["bill_date"] = pd.to_datetime(
    billing["bill_date"])
```

```python
#Standardize text
#columns name
print("Patients")
print(patients.select_dtypes(include='object').columns)

print("\nDoctors")
print(doctors.select_dtypes(include='object').columns)

print("\nAppointments")
print(appointments.select_dtypes(include='object').columns)

print("\nTreatments")
print(treatments.select_dtypes(include='object').columns)

print("\nBilling")
print(billing.select_dtypes(include='object').columns)
```

```python
#Patients Dataset
# Standardize names
patients["first_name"] = patients["first_name"].str.title().str.strip()
patients["last_name"] = patients["last_name"].str.title().str.strip()

# Standardize gender
patients["gender"] = patients["gender"].str.title().str.strip()

# Standardize address
patients["address"] = patients["address"].str.title().str.strip()

# Standardize insurance provider
patients["insurance_provider"] = patients["insurance_provider"].str.title().str.strip()

# Emails should be lowercase
patients["email"] = patients["email"].str.lower().str.strip()
```

```python
#Appointments Dataset
# Standardize reason for visit
appointments["reason_for_visit"] = appointments["reason_for_visit"].str.title().str.strip()

# Standardize appointment status
appointments["status"] = appointments["status"].str.title().str.strip()
```

```python
#Treatments Dataset
# Standardize treatment type
treatments["treatment_type"] = treatments["treatment_type"].str.title().str.strip()

# Standardize description
treatments["description"] = treatments["description"].str.title().str.strip()
```

```python
#billing dataset
# Standardize payment method
billing["payment_method"] = billing["payment_method"].str.title().str.strip()

# Standardize payment status
billing["payment_status"] = billing["payment_status"].str.title().str.strip()
```

```python
#checking unique value
print(patients["gender"].unique())
print(doctors["specialization"].unique())
print(appointments["status"].unique())
print(treatments["treatment_type"].unique())
print(billing["payment_status"].unique())
```

```python
#Data Integration (Merge the Datasets)
patient_appointments = pd.merge(
    patients,
    appointments,
    on="patient_id",
    how="inner"
)

patient_appointments.head()
```

```python
patient_doctor = pd.merge(
    patient_appointments,
    doctors,
    on="doctor_id",
    how="inner"
)

patient_doctor.head()
```

```python
patient_treatment = pd.merge(
    patient_doctor,
    treatments,
    on="appointment_id",
    how="inner"
)

patient_treatment.head()
```

```python
healthcare_data = pd.merge(
    patient_treatment,
    billing,
    on=["patient_id", "treatment_id"],
    how="inner"
)

healthcare_data.head()
```
