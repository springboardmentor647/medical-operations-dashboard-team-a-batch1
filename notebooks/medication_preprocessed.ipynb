```python
df["Medicine"] = df["Medicine"].str.strip()
```

```python
df["Medicine"] = df["Medicine"].str.title()
```

```python
df["Admission_ID"].nunique()
```

```python
dosage_map = {
    "Paracetamol": ["500 mg", "650 mg"],
    "Ibuprofen": ["200 mg", "400 mg", "600 mg"],
    "Insulin": ["5 Units", "10 Units", "15 Units"]
}

df["Dosage"] = df["Medicine"].apply(lambda x: random.choice(dosage_map[x]))
```

```python
frequency_map = {
    "Paracetamol": ["Twice Daily", "Three Times Daily"],
    "Ibuprofen": ["Twice Daily", "Three Times Daily"],
    "Insulin": ["Once Daily", "Before Meals"]
}

df["Frequency"] = df["Medicine"].apply(lambda x: random.choice(frequency_map[x]))
```

```python
cost_map = {
    "Paracetamol": (10, 30),
    "Ibuprofen": (20, 60),
    "Insulin": (350, 900)
}

df["Cost"] = df["Medicine"].apply(
    lambda x: round(random.uniform(*cost_map[x]), 2)
)
```

```python
df["Prescribed_By"] = [
    f"D{random.randint(1,350):05d}"
    for _ in range(len(df))
]
```

```python
start_date = datetime(2024, 1, 1)
end_date = datetime(2025, 12, 31)

df["Sell_Date"] = [
    start_date + timedelta(days=random.randint(0, (end_date - start_date).days))
    for _ in range(len(df))
]

df["Sell_Date"] = df["Sell_Date"].dt.strftime("%Y-%m-%d")
```

```python
manufacturer_map = {
    "Paracetamol": ["Sun Pharma", "Cipla", "Lupin"],
    "Ibuprofen": ["Dr. Reddy's", "Sun Pharma", "Torrent Pharma"],
    "Insulin": ["Novo Nordisk", "Sanofi", "Eli Lilly"]
}

df["Manufacturer"] = df["Medicine"].apply(
    lambda x: random.choice(manufacturer_map[x])
)
```
