```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
master = pd.read_csv("Hospital_Master_Data_Final.csv")
```

```python
master.head()
```

```python
#check columns
print(master.columns.tolist())
```

```python
master["Admission_Date"].head()
```

```python
print(master["Admission_Date"].dtype)
```

```python
# Step 2: Convert Admission_Date from text format to datetime format.
# Datetime conversion is required for extracting year, month, and quarter
# and for performing time-based admission trend analysis.
```

```python
master["Admission_Date"] = pd.to_datetime(
    master["Admission_Date"],
    errors="coerce"
)
```

```python
print(master["Admission_Date"].isnull().sum())
```

```python
print("Earliest Admission:", master["Admission_Date"].min())
print("Latest Admission:", master["Admission_Date"].max())
#date range
```

```python
# Step 3: Extract the admission year to support yearly admission analysis.
master["Admission_Year"] = master["Admission_Date"].dt.year
```

```python
print(master["Admission_Year"].value_counts().sort_index())
```

```python
# Step 4: Extract the admission month and month name
# to analyze monthly variations in patient admissions.
```

```python
master["Admission_Month"] = master["Admission_Date"].dt.month
```

```python
#creating proper month name
master["Admission_Month_Name"] = master["Admission_Date"].dt.month_name()
```

```python
master[["Admission_Date", "Admission_Month", "Admission_Month_Name"]].head()
```

```python
# Step 5: Create a Month-Year field to preserve the chronological order
# of admissions across different years.
```

```python
master["Admission_Month_Year"] = (
    master["Admission_Date"]
    .dt.to_period("M")
    .astype(str)
)
```

```python
master[
    ["Admission_Date", "Admission_Month_Year"]
].head()
```

```python
# Step 6: Create admission quarter information
# to analyze quarterly changes in admission volume.
```

```python
master["Admission_Quarter"] = (
    master["Admission_Date"]
    .dt.to_period("Q")
    .astype(str)
)
```

```python
master[
    ["Admission_Date", "Admission_Quarter"]
].head()
```

```python
# Step 7: Calculate the total number of unique admissions.
# Unique Admission_ID values are used to avoid counting the same admission multiple times.
```

```python
total_admissions = master["Admission_ID"].nunique()

print("Total Admissions:", total_admissions)
```

```python
# Step 8: Calculate the number of unique admissions for each month.
# This dataset will be used to analyze monthly admission trends.
```

```python
monthly_admissions = (
    master.groupby("Admission_Month_Year")["Admission_ID"]
    .nunique()
    .reset_index(name="Admission_Count")
)
```

```python
monthly_admissions
```

```python
# Step 9: Sort monthly admission records chronologically
# so that the trend is displayed in the correct time sequence.
```

```python
monthly_admissions = monthly_admissions.sort_values(
    "Admission_Month_Year"
)
```

```python
# Step 10: Identify the month with the highest admission volume.
# This helps identify periods of peak patient demand.
```

```python
highest_month = monthly_admissions.loc[
    monthly_admissions["Admission_Count"].idxmax()
]

print(highest_month)
```

```python
# Step 11: Identify the month with the lowest admission volume.
# This helps identify periods of relatively lower patient demand.
```

```python
lowest_month = monthly_admissions.loc[
    monthly_admissions["Admission_Count"].idxmin()
]

print(lowest_month)
```

```python
# Step 12: Visualize the monthly admission trend using a line chart.
# The visualization helps identify increases, decreases, peaks,
# and fluctuations in patient admission volume over time.
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    monthly_admissions["Admission_Month_Year"],
    monthly_admissions["Admission_Count"],
    marker="o"
)

plt.xlabel("Month")
plt.ylabel("Number of Admissions")
plt.title("Monthly Admission Trend")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Step 13: Calculate the number of unique admissions for each quarter
# to identify broader periodic variations in patient demand.
```

```python
quarterly_admissions = (
    master.groupby("Admission_Quarter")["Admission_ID"]
    .nunique()
    .reset_index(name="Admission_Count")
)
```

```python
quarterly_admissions
```

```python
#sort quarterly data
quarterly_admissions = quarterly_admissions.sort_values(
    "Admission_Quarter"
)
```

```python
# Step 14: Visualize quarterly admission volumes using a bar chart.
# The chart provides a higher-level view of patient admission patterns
# across quarters.
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    quarterly_admissions["Admission_Quarter"],
    quarterly_admissions["Admission_Count"]
)

plt.xlabel("Quarter")
plt.ylabel("Number of Admissions")
plt.title("Quarterly Admission Trend")

plt.tight_layout()
plt.show()
```

```python
# Step 15: Calculate unique admissions by department
# to identify departments with higher patient demand and workload.
```

```python
print(master["Department_Name"].unique())
```

```python
print(master["Department_Name"].isnull().sum())
```

```python
department_admissions = (
    master.groupby("Department_Name")["Admission_ID"]
    .nunique()
    .reset_index(name="Admission_Count")
    .sort_values("Admission_Count", ascending=False)
)
```

```python
department_admissions
```

```python
# Step 16: Visualize department-wise admission volume.
# This helps identify departments receiving higher numbers of patients
# and supports workload and resource planning.
```

```python
plt.figure(figsize=(10, 6))

plt.barh(
    department_admissions["Department_Name"],
    department_admissions["Admission_Count"]
)

plt.xlabel("Number of Admissions")
plt.ylabel("Department")
plt.title("Department-wise Admission Volume")

plt.tight_layout()
plt.show()
```

```python
# Step 17: Examine the distribution of admission statuses
# to understand the different outcomes recorded in the dataset.
```

```python
print(master["Status"].value_counts())
```

```python
status_admissions = (
    master.groupby("Status")["Admission_ID"]
    .nunique()
    .reset_index(name="Admission_Count")
    .sort_values("Admission_Count", ascending=False)
)

status_admissions
```

```python
status_admissions = (
    master.groupby("Status")["Admission_ID"]
    .nunique()
    .reset_index(name="Admission_Count")
    .sort_values("Admission_Count", ascending=False)
)

status_admissions
```

```python
# Step 18: Visualize the distribution of admission statuses.
# This helps identify the proportion of different admission outcomes
# recorded in the healthcare dataset.
```

```python
plt.figure(figsize=(9, 5))

plt.bar(
    status_admissions["Status"],
    status_admissions["Admission_Count"]
)

plt.xlabel("Admission Status")
plt.ylabel("Number of Admissions")
plt.title("Admission Status Distribution")

plt.tight_layout()
plt.show()
```

```python
# Step 19: Calculate monthly admission volumes for each department.
# This allows department-level patient demand to be monitored over time.
```

```python
monthly_department = (
    master.groupby(
        ["Admission_Month_Year", "Department_Name"]
    )["Admission_ID"]
    .nunique()
    .reset_index(name="Admission_Count")
)

monthly_department.head()
```

```python
# Step 20: Visualize monthly admission trends for each department.
# This helps compare changes in patient demand across departments over time.
```

```python
plt.figure(figsize=(12, 6))

for department in monthly_department["Department_Name"].unique():

    data = monthly_department[
        monthly_department["Department_Name"] == department
    ]

    plt.plot(
        data["Admission_Month_Year"],
        data["Admission_Count"],
        marker="o",
        label=department
    )

plt.xlabel("Month")
plt.ylabel("Number of Admissions")
plt.title("Monthly Admission Trend by Department")
plt.xticks(rotation=45)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# Step 21: Identify the department with the highest admission volume
# to highlight areas with relatively greater patient demand.
```

```python
highest_department = department_admissions.iloc[0]

print("Department with highest admissions:")
print(highest_department)
```
