```python
import pandas as pd

patients = pd.read_csv("patients_clean.csv")
admissions = pd.read_csv("admissions_clean.csv")
doctors = pd.read_csv("doctors_preprocessed.csv")
departments = pd.read_csv("departments_clean.csv")
medications = pd.read_csv("medications (1).csv")
billing = pd.read_csv("billing_clean.csv")
lab = pd.read_csv("lab_results_clean.csv")
surgeries = pd.read_csv("surgeries_clean_fixed.csv")
```

```python
departments.rename(columns={
    "department_id":"Department_ID",
    "department_name":"Department_Name"
}, inplace=True)
```

```python
#Start with Admissions
master = admissions.copy()
```

```python
#merge patients using Patient_ID
master = master.merge(
    patients,
    on="Patient_ID",
    how="left"
)
```

```python
#merge doctors using Doctor_ID
master = master.merge(
    doctors,
    on="Doctor_ID",
    how="left"
)
```

```python
#Clean duplicate Department_ID
master.drop(columns=["Department_ID_y"], inplace=True)
master.rename(columns={
    "Department_ID_x": "Department_ID"
}, inplace=True)
```

```python
#merge department using department_id
master = master.merge(
    departments,
    on="Department_ID",
    how="left"
)
```

```python

#master = master.merge(
 #   doctors.drop(columns=["Department_ID"]),
  #  on="Doctor_ID",
   # how="left"
#)
```

```python
#merge billing
master = master.merge(
    billing,
    on=["Admission_ID","Patient_ID"],
    how="left"
)
```

```python
#merge lab result
master = master.merge(
    labs,
    on="Admission_ID",
    how="left"
)
```

```python
#merge medication
master = master.merge(
    medications,
    on="Admission_ID",
    how="left"
)
```

```python
#merge surgeries
master = master.merge(
    surgeries,
    on=["Admission_ID","Patient_ID","Doctor_ID"],
    how="left"
)
```

```python
#reomove duplicate columns
#(to remove Doctor_ID_x Doctor_ID_y Patient_ID_x Patient_ID_y
master.rename(columns={
    "Doctor_ID_x":"Doctor_ID",
    "Patient_ID_x":"Patient_ID"
}, inplace=True)

master.drop(
    columns=[
        "Doctor_ID_y",
        "Patient_ID_y"
    ],
    errors="ignore",
    inplace=True
)
```

```python
#convert dates
date_cols = [
    "Admission_Date",
    "Discharge_Date",
    "Sell_Date",
    "Surgery_Date"
]

for col in date_cols:
    if col in master.columns:
        master[col] = pd.to_datetime(master[col])
```
